```python
# Cell 1 — Imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Cell 2 — Load processed data
df = pd.read_csv("../data/processed/cc_segmentation_raw.csv")
print("Shape:", df.shape)
df.head()
```

```python
# Cell 3 — Basic statistics
df.describe().T
```

```python
# Cell 4 — Distribution of all features
df.hist(bins=50, figsize=(20, 15))
plt.suptitle("Feature Distributions", y=1.02, fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Cell 5 — Correlation heatmap
plt.figure(figsize=(16, 12))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
# Cell 6 — Outlier check using boxplots
plt.figure(figsize=(20, 10))
df.boxplot(rot=45)
plt.title("Boxplot — Outlier Overview")
plt.tight_layout()
plt.show()
```

```python
# Cell 1 — Imports
import pandas as pd
```

```python
# Cell 2 — Load dataset
df = pd.read_csv("../data/raw/CC GENERAL.csv")
```

```python
# Cell 3 — First look
print("Shape:", df.shape)
df.head()
```

```python
# Cell 4 — Missing values
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct})
```

```python
# Cell 5 — Duplicates
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Cell 6 — Drop CUST_ID
df = df.drop(columns=["CUST_ID"])
print("Shape after dropping CUST_ID:", df.shape)
```

```python
# Cell 7 — Save cleaned file
df.to_csv("../data/processed/cc_segmentation_raw.csv", index=False)
print("Saved to data/processed/cc_segmentation_raw.csv")
```

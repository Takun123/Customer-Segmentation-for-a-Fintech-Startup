```python
# Cell 1 — Imports
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
# Cell 2 — Load data
df = pd.read_csv("../data/processed/cc_segmentation_raw.csv")
print("Shape:", df.shape)
```

```python
# Cell 3 — Impute MINIMUM_PAYMENTS with median, drop CREDIT_LIMIT null row
df["MINIMUM_PAYMENTS"] = df["MINIMUM_PAYMENTS"].fillna(df["MINIMUM_PAYMENTS"].median())
df = df.dropna(subset=["CREDIT_LIMIT"])
print("Shape after handling nulls:", df.shape)
print("Remaining nulls:", df.isnull().sum().sum())
```

```python
# Cell 4 — Cap outliers at 95th percentile
cols_to_cap = [
    "BALANCE", "PURCHASES", "ONEOFF_PURCHASES", "INSTALLMENTS_PURCHASES",
    "CASH_ADVANCE", "PURCHASES_TRX", "CASH_ADVANCE_TRX",
    "PAYMENTS", "MINIMUM_PAYMENTS", "CREDIT_LIMIT"
]

for col in cols_to_cap:
    cap = df[col].quantile(0.95)
    df[col] = df[col].clip(upper=cap)

print("Outlier capping complete.")
```

```python
# Cell 5 — Scale all features
scaler = StandardScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)
print("Scaling complete.")
df_scaled.describe().T.round(2)
```

```python
# Cell 6 — Save processed file
df_scaled.to_csv("../data/processed/cc_segmentation_processed.csv", index=False)
print("Saved to data/processed/cc_segmentation_processed.csv")
```
